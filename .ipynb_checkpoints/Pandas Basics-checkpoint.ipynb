```python
import numpy as np

df = pd.DataFrame({
    "Name":["A","B","C","D","E"],
    "Age":[20,21,22,np.nan,24],
    "Marks":[80,90,np.nan,85,95]
})
```

```python
print(df[(df["Age"]>20) & (df["Marks"]>85)])
```

```python
df.isnull().sum()
```
